```python
import pandas as pd
import numpy as np
cancer_data=pd.read_csv(
    "../data/structured/processed/cancer_feature_engineered.csv"
)
cancer_data.head()
```

```python
cancer_data.info()
```

## Separate X and y target is Target_Severity_score

```python
X=cancer_data.drop(
    columns=["Target_Severity_Score"]
)
y=cancer_data["Target_Severity_Score"]
```

## Identify Categorical Columns

```python
categorical_cols=X.select_dtypes(
    include=["object","category"]
).columns
categorical_cols
```

## One-Hot Encoding

```python
X_encoded=pd.get_dummies(
    X,
    columns=categorical_cols,
    drop_first=True
)
```

```python
X_encoded.shape
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
scalar=StandardScaler()
X_scaled=scalar.fit_transform(X_encoded)
```

```python
import pandas as pd
X_scaled=pd.DataFrame(
    X_scaled,
    columns=X_encoded.columns
)
X_scaled.head()
```

## Train-Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test=train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
X_scaled.to_csv(
    "../data/structured/processed/X_preprocessed.csv",
    index=False
)
```
