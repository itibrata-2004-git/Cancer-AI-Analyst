## Import Libraries and Load Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
cancer_data=pd.read_csv(r"F:\Cancer AI Analyst\data\structured\processed\cancer_cleaned.csv"
)
```

## Age Distribution

```python
plt.figure(figsize=(10,5))
sns.histplot(
    cancer_data["Age"],
    bins=20,
    kde=True,
    color="royalblue"
)
plt.title(
    "Distribution of Patient Age",
    fontsize=14,
    fontweight="bold"
)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## Target Severity Score Distribution

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.set_style("darkgrid")
sns.set_palette("Set2")

sns.histplot(
    cancer_data["Target_Severity_Score"].dropna(),
    bins=20,
    kde=True,
    color="teal"
)

plt.title("Target Severity Score Distribution")
plt.xlabel("Target Severity Score")
plt.ylabel("Frequency")
plt.show()
```

## Cancer Type Distribution

```python
plt.figure(figsize=(10,5))
sns.countplot(
    data=cancer_data,
    x="Cancer_Type",
    color="skyblue"
)
plt.xticks(rotation=45)
plt.show()
```

## Cancer Stage Distribution

```python
plt.figure(figsize=(8,5))
sns.countplot(
    data=cancer_data,
    x="Cancer_Stage",
    color="lightgreen"
)
plt.show()
```

## Smoking Vs Severity Score

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=cancer_data,
    x="Smoking",
    y="Target_Severity_Score"
)
plt.title(
    "Smoking vs Severity Score",
    fontsize=14,
    fontweight="bold"
)
plt.show()
```

## Obesity VS Severity Score

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=cancer_data,
    x="Obesity_Level",
    y="Target_Severity_Score"
)
plt.title(
    "Obesity Level vs Severity Score",
    fontsize=14,
    fontweight="bold"
)
plt.show()
```

## Genetic Risk vs Severity Score

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=cancer_data,
    x="Genetic_Risk",
    y="Target_Severity_Score"
)
plt.title(
    "Genetic Risk vs Severity Score",
    fontsize=14,
    fontweight="bold"
)
plt.show()
```

## Cancer Stage vs Severity Score

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    data=cancer_data,
    x="Cancer_Stage",
    y="Target_Severity_Score"
)
plt.title(
    "Cancer Stage vs Severity Score",
    fontsize=14,
    fontweight="bold"
)
plt.show()
```

## Cancer Type vs Treatment Cost

```python
plt.figure(figsize=(12,5))
sns.boxplot(
    data=cancer_data,
    x="Cancer_Type",
    y="Treatment_Cost_USD"
)
plt.xticks(rotation=45)
plt.title(
    "Cancer Type vs Treatment Cost",
    fontsize=14,
    fontweight="bold"
)
plt.show()
```

## Correlation Heatmap

```python
plt.figure(figsize=(10,8))
corr_matrix = cancer_data.select_dtypes(
    include=["int64","float64"]
).corr()
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title(
    "Correlation Heatmap"
)
plt.show()
```

## Pairplot

```python
eda_features = [
    "Age",
    "Smoking",
    "Obesity_Level",
    "Genetic_Risk",
    "Target_Severity_Score"
]
sns.pairplot(
    cancer_data[eda_features],
    corner=True
)
plt.show()
```

## Severity Score Across Multiple Risk Factor

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=cancer_data,
    x="Smoking",
    y="Target_Severity_Score",
    hue="Cancer_Stage"
)
plt.title(
    "Smoking vs Severity by Cancer Stage"
)
plt.show()
```

## Treatment Cost Analysis

```python
plt.figure(figsize=(10,6))
sns.boxplot(
    data=cancer_data,
    x="Cancer_Stage",
    y="Treatment_Cost_USD"
)
plt.title(
    "Treatment Cost by Cancer Stage"
)
plt.show()
```

## Country-wise Cancer Analysis

```python
country_counts=(
    cancer_data["Country_Region"]
    .value_counts()
    .head(10)
)
plt.figure(figsize=(10,5))
sns.barplot(
    x=country_counts.index,
    y=country_counts.values
)
plt.xticks(rotation=45)
plt.title(
    "Top 10 Countries by Patient Count"
)
plt.show()
```

## Correlation with Target Severity Score and create Bar Plot to show them

```python
corr=(
    cancer_data
    .select_dtypes(
        include=["int64","float64"]
    )
    .corr()
)
target_corr = (
    corr["Target_Severity_Score"]
    .sort_values(
        ascending=False
    )
)
print(target_corr)
```

```python
plt.figure(figsize=(8,5))
target_corr.drop(
    "Target_Severity_Score"
).plot(
    kind="bar"
)
plt.title(
    "Feature Correlation with Severity Score"
)
plt.ylabel("Correlation")
plt.show()
```
