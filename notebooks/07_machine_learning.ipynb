## Load Train-Test  Split Data

```python
import joblib
X_train, X_test, y_train, y_test = joblib.load(
    "../models/train_test_split.pkl"
)
```

## Import Evaluation Metrics

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

## Create Evaluation Function

```python
import numpy as np
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
def evaluate_model(
    model,
    X_test,
    y_test
):

    predictions=model.predict(X_test)

    mae=mean_absolute_error(
        y_test,
        predictions
    )

    mse=mean_squared_error(
        y_test,
        predictions
    )

    rmse=np.sqrt(mse)

    r2=r2_score(
        y_test,
        predictions
    )

    return {
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2
    }
```

## Model 1: Linear Regression

```python
import pandas as pd
X_train = pd.get_dummies(X_train, drop_first=True)
X_test = pd.get_dummies(X_test, drop_first=True)

X_train, X_test = X_train.align(
    X_test,
    join="left",
    axis=1,
    fill_value=0
)
```

```python
print(X_train.isnull().sum())
```

```python
print(y_train.isnull().sum())
```

```python
from sklearn.impute import SimpleImputer

imputer=SimpleImputer(strategy="most_frequent")

X_train["Stage_Severity"] = imputer.fit_transform(
    X_train[["Stage_Severity"]]
).ravel()

X_test["Stage_Severity"]=imputer.transform(
    X_test[["Stage_Severity"]]
).ravel()
```

```python
X_train=pd.get_dummies(X_train, drop_first=True)
X_test=pd.get_dummies(X_test, drop_first=True)

X_train, X_test=X_train.align(
    X_test,
    join="left",
    axis=1,
    fill_value=0
)
```

```python
print(X_train.isnull().sum().sum())
print(X_test.isnull().sum().sum())
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(X_train, y_train)
```

```python
lr_results=evaluate_model(
    lr,
    X_test,
    y_test
)
lr_results
```

## Model 2: Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf=RandomForestRegressor(
    n_estimators=200,
    random_state=42
)
rf.fit(
    X_train,
    y_train
)
```

```python
rf_results=evaluate_model(
    rf,
    X_test,
    y_test
)
rf_results
```

## Model 3: Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
gb=GradientBoostingRegressor(
    random_state=42
)
gb.fit(
    X_train,
    y_train
)
```

```python
gb_results=evaluate_model(
    gb,
    X_test,
    y_test
)
gb_results
```

## Model 4: XGBoost

```python
from xgboost import XGBRegressor
```

```python
xgb=XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    random_state=42
)
xgb.fit(
    X_train,
    y_train
)
```

```python
xgb_results=evaluate_model(
    xgb,
    X_test,
    y_test
)
xgb_results
```

## Compare the 4 Models and took best one

```python
results=pd.DataFrame({
    "Linear Regression": lr_results,
    "Random Forest": rf_results,
    "Gradient Boosting": gb_results,
    "XGBoost": xgb_results
}).T
results
```
