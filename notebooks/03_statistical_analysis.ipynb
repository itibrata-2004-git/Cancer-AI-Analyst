## Load the Saved Datasets

```python
import pandas as pd
import numpy as np
```

```python
cancer_data=pd.read_csv(r'F:\Cancer AI Analyst\data\structured\processed\cancer_cleaned.csv')
cancer_data.head()
```

## Separate Numerical and Categorical Columns

```python
numerical_cols=cancer_data.select_dtypes(
    include=["int64", "float64"]
).columns
categorical_cols=cancer_data.select_dtypes(
    include=["object"]
).columns
print("Numerical Columns:")
print(numerical_cols)
print("\nCategorical Columns:")
print(categorical_cols)
```

## Descriptive Statistics

```python
cancer_data[numerical_cols].describe().T
```

## Central Tendency Analysis

```python
central_tendency=pd.DataFrame({
    "Mean": cancer_data[numerical_cols].mean(),
    "Median": cancer_data[numerical_cols].median(),
    "Mode": cancer_data[numerical_cols].mode().iloc[0]
})
central_tendency
```

## Dispersion Analysis

```python
dispersion=pd.DataFrame({
    "Range": cancer_data[numerical_cols].max()
             - cancer_data[numerical_cols].min(),
    "Variance": cancer_data[numerical_cols].var(),
    "Std_Dev": cancer_data[numerical_cols].std()
})
dispersion.sort_values(
    by="Std_Dev",
    ascending=False
)
```

## Skewness Analysis

```python
skewness=cancer_data[numerical_cols].skew()
skewness.sort_values(
    ascending=False
)
```

## Kurtosis Analysis

```python
kurtosis=cancer_data[numerical_cols].kurtosis()
kurtosis.sort_values(
    ascending=False
)
```

## Categorical Analysis

```python
for col in categorical_cols:
    print("\n" + "="*50)
    print(col)
    print(
        cancer_data[col].value_counts()
    )
```

## Correlation Analysis

```python
correlation_matrix=cancer_data[
    numerical_cols
].corr()
correlation_matrix
```

## Target-Based Correlation

```python
correlation_matrix[
    "Target_Severity_Score"
].sort_values(
    ascending=False
)
```
